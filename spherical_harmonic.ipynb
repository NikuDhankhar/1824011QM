```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import lpmv
from math import factorial
```

```python
def spherical_harmonic(quantum_numbers):
    # Angular coordinates
    theta = np.linspace(0, np.pi, 100)
    phi = np.linspace(0, 2*np.pi, 100)

    theta, phi = np.meshgrid(theta, phi)

    # Create figure
    fig = plt.figure(figsize=(15, 12))

    for i, (l, m) in enumerate(quantum_numbers):

        # Normalization constant
        N = np.sqrt((2*l + 1)/(4*np.pi) *
                factorial(l-m)/factorial(l+m))

        #Legendre polynomial
        P = lpmv(m, l, np.cos(theta))

        # Spherical harmonic
        Y = N * P * np.exp(1j*m*phi)

        # Probability density
        prob = np.abs(Y)**2

        # Cartesian coordinates
        x = prob * np.sin(theta) * np.cos(phi)
        y = prob * np.sin(theta) * np.sin(phi)
        z = prob * np.cos(theta)

        # 3D plot
        ax = fig.add_subplot(4, 3, i+1, projection='3d')
        ax.plot_surface(x, y, z, cmap='viridis')
        ax.set_title(f'l = {l}, m = {m}')

plt.tight_layout()
plt.show()
```

```python
quantum_numbers = [(2,)]
spherical_harmonic(quantum_numbers)
```
